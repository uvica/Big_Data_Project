```python
import os
import xml.etree.ElementTree as ET
import pandas as pd
```

```python
folder = "../data/timetable/extracted"

namespace = {
    "tx": "http://www.transxchange.org.uk/"
}

records = []
```

```python
for file in os.listdir(folder):

    if not file.endswith(".xml"):
        continue

    filepath = os.path.join(folder, file)

    try:

        tree = ET.parse(filepath)
        root = tree.getroot()

        operator = root.find(
            ".//tx:OperatorShortName",
            namespace
        )

        if operator is None:
            operator = root.find(
                ".//tx:OperatorNameOnLicence",
                namespace
            )


        service = root.find(
            ".//tx:ServiceCode",
            namespace
        )


        journeys = root.findall(
            ".//tx:VehicleJourney",
            namespace
        )


        for journey in journeys:

            journey_code = journey.find(
                "tx:VehicleJourneyCode",
                namespace
            )


            timing_links = journey.findall(
                "tx:VehicleJourneyTimingLink",
                namespace
            )


            for sequence, link in enumerate(timing_links):

                records.append({

                    "file": file,

                    "operator":
                    operator.text if operator is not None else None,

                    "service_code":
                    service.text if service is not None else None,

                    "journey_id":
                    journey_code.text if journey_code is not None else None,

                    "stop_sequence":
                    sequence + 1

                })


    except Exception as e:
        print("Error:", file, e)
```

```python
print("Total records:", len(records))
```
